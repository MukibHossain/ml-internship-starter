# 5. Rule-Based Systems

## Definition

Rule-based systems solve problems using predefined IF–THEN rules rather than learning from data.

## Advantages

- Easy to understand
- Fast execution
- No training data required
- Predictable behavior

## Limitations

- Cannot learn new patterns
- Difficult to maintain with many rules
- Poor performance on complex problems

## Example

IF Age ≥ 18

THEN Eligible to Vote

# 6. AI vs ML vs Analytics vs Rule-Based Systems

- **Artificial Intelligence (AI):** A broad field focused on creating systems that can perform tasks requiring human intelligence.
- **Machine Learning (ML):** A subset of AI that learns patterns from data to make predictions or decisions.
- **Analytics:** The process of examining data to gain insights and support decision-making.
- **Rule-Based Systems:** Systems that follow predefined IF–THEN rules without learning from data.

# 7. When Should We Use Machine Learning?

Machine Learning should be used when:

- Large amounts of data are available.
- Relationships between variables are complex.
- The problem requires prediction.
- Rules cannot capture all possible patterns.

## Examples

- Disease prediction
- Face recognition
- Customer recommendation
- Credit risk prediction
- Traffic forecasting

# 8. When Is a Rule Better Than Machine Learning?

Rule-based systems are better when:

- The logic is simple.
- Rules rarely change.
- No historical data exists.
- Complete transparency is required.

## Examples

- Tax calculation
- Voting eligibility
- Password validation
- Attendance calculation

# 10. Supervised Learning

# 11. Unsupervised Learning

## Definition

Unsupervised learning discovers hidden patterns in data without using labels.

## Characteristics

- No target variable
- Finds natural groups

## Examples

- Customer segmentation
- Document clustering
- Anomaly detection

## Advantages

- No labeled data required

## Limitations

- Harder to evaluate
- Results may be difficult to interpret

# 12. Supervised vs Unsupervised Learning

### Supervised Learning
Supervised learning uses **labeled data**, where the target variable is known. Its goal is to learn the relationship between input features and the target so it can make accurate predictions. Common supervised learning tasks include **classification** and **regression**.

### Unsupervised Learning
Unsupervised learning uses **unlabeled data**, where no target variable is provided. Its goal is to discover hidden patterns, structures, or relationships within the data. A common unsupervised learning task is **clustering**.

# 13. Generalization

Generalization is the ability of a machine learning model to perform well on new, unseen data. A model that generalizes effectively learns meaningful patterns rather than memorizing the training data, allowing it to make reliable predictions in real-world scenarios.

# 14. Overfitting

Overfitting occurs when a model learns the training data too closely, including noise and irrelevant details. As a result, it performs well on training data but poorly on unseen data.

## Prevention

- Cross-validation
- Regularization
- Early stopping
- More training data
- Simpler models

# 15. Generalization vs Overfitting

### Generalization

Generalization means a machine learning model has learned the important patterns from the training data and can make accurate predictions on new, unseen data. A model with good generalization performs well in real-world situations.

### Overfitting

Overfitting happens when a model memorizes the training data instead of learning the underlying patterns. As a result, it performs very well on the training data but poorly on new, unseen data.
